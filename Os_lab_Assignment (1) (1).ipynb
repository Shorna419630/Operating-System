```python

# FCFS Scheduling with Average Waiting Time

processes = ["P1", "P2", "P3", "P4"]
burst_time = [21, 3, 6, 2]

n = len(processes)

waiting_time = [0] * n
completion_time = [0] * n

# Waiting time calculation
for i in range(1, n):
    waiting_time[i] = waiting_time[i-1] + burst_time[i-1]

# Completion time calculation
completion_time[0] = burst_time[0]
for i in range(1, n):
    completion_time[i] = completion_time[i-1] + burst_time[i]

# Average waiting time
avg_waiting = sum(waiting_time) / n

print("Process\tBurst Time")
for i in range(n):
    print(f"{processes[i]}\t{burst_time[i]}")

print("\nGantt Chart:")

# Gantt Chart bar
print("|", end="")
for p in processes:
    print(f"  {p}  |", end="")
print()

# Time line
time = 0
print(time, end="\t")
for bt in burst_time:
    time += bt
    print(time, end="\t")

print("\n\nAverage Waiting Time =", avg_waiting, "ms")
```

```python

# SJF Scheduling (Process order same as input)

processes = ["P1", "P2", "P3", "P4"]
burst_time = [21, 3, 6, 2]

n = len(processes)

print("Process\tBurst Time")
for i in range(n):
    print(processes[i], "\t", burst_time[i])

# SJF calculation
data = list(zip(processes, burst_time))
data.sort(key=lambda x: x[1])   # sort by burst time

sjf_process = [x[0] for x in data]
sjf_burst = [x[1] for x in data]

waiting_time = [0]*n
completion_time = [0]*n

for i in range(1, n):
    waiting_time[i] = waiting_time[i-1] + sjf_burst[i-1]

completion_time[0] = sjf_burst[0]
for i in range(1, n):
    completion_time[i] = completion_time[i-1] + sjf_burst[i]

avg_wait = sum(waiting_time)/n

print("\nGantt Chart:")

print("|", end="")
for p in sjf_process:
    print(f"  {p}  |", end="")
print()

time = 0
print(time, end="\t")
for bt in sjf_burst:
    time += bt
    print(time, end="\t")

print("\n\nAverage Waiting Time =", avg_wait, "ms")
```

```python

# Priority Scheduling (Non-Preemptive)

processes = ["P1", "P2", "P3", "P4"]
burst_time = [21, 3, 6, 2]
priority = [2, 1, 4, 3]

n = len(processes)

print("Process\tBurst Time\tPriority")
for i in range(n):
    print(processes[i], "\t", burst_time[i], "\t\t", priority[i])

# combine data
data = list(zip(processes, burst_time, priority))

# sort by priority (lower number = higher priority)
data.sort(key=lambda x: x[2])

p = [x[0] for x in data]
b = [x[1] for x in data]

waiting_time = [0]*n
completion = [0]*n

for i in range(1, n):
    waiting_time[i] = waiting_time[i-1] + b[i-1]

completion[0] = b[0]
for i in range(1, n):
    completion[i] = completion[i-1] + b[i]

avg_wait = sum(waiting_time)/n

print("\nGantt Chart:")

print("|", end="")
for i in p:
    print(f"  {i}  |", end="")
print()

time = 0
print(time, end="\t")
for i in b:
    time += i
    print(time, end="\t")

print("\n\nAverage Waiting Time =", avg_wait, "ms")
```

```python

# Round Robin Scheduling

process = ["P1", "P2", "P3", "P4"]
burst = [21, 3, 6, 2]
quantum = 5

remaining = burst.copy()
time = 0
gantt = []
times = [0]
completion = [0]*len(process)

while True:
    done = True
    for i in range(len(process)):
        if remaining[i] > 0:
            done = False
            gantt.append(process[i])

            if remaining[i] > quantum:
                time += quantum
                remaining[i] -= quantum
            else:
                time += remaining[i]
                completion[i] = time
                remaining[i] = 0

            times.append(time)

    if done:
        break

# Waiting Time Calculation
waiting = []
for i in range(len(process)):
    waiting.append(completion[i] - burst[i])

avg_wait = sum(waiting) / len(process)

# Process Table
print("Process   Burst Time")
for i in range(len(process)):
    print(process[i], "      ", burst[i])

# Gantt Chart
print("\nGantt Chart:\n")

print("|", end=" ")
for p in gantt:
    print(p, "|", end=" ")

print("\n")

for t in times:
    print(t, "   ", end="")

print("\n\nAverage Waiting Time =", avg_wait)
```

```python
#FIFO
def fifo_page_replacement(pages, frames):
    memory = []
    page_faults = 0
    index = 0

    print("Ref\tFrame1\tFrame2\tFrame3\tStatus")

    for page in pages:
        if page not in memory:
            status = "Fault"
            page_faults += 1

            if len(memory) < frames:
                memory.append(page)
            else:
                memory[index] = page
                index = (index + 1) % frames
        else:
            status = "Hit"

        # frame print
        frame_state = memory + ["-"] * (frames - len(memory))
        print(page, "\t", frame_state[0], "\t", frame_state[1], "\t", frame_state[2], "\t", status)

    print("\nTotal Page Faults =", page_faults)


# reference string
pages = [7,0,1,2,0,3,0,4,2,3,0,3,2,1,2,0,1,7,0,1]
frames = 3

fifo_page_replacement(pages, frames)
```

```python

# Optimal Page Replacement

def optimal_page_replacement(pages, frames):

    frame = []
    faults = 0

    print("Ref\tFrame1\tFrame2\tFrame3\tStatus")

    for i in range(len(pages)):

        page = pages[i]

        if page in frame:
            status = "Hit"

        else:
            faults += 1
            status = "Fault"

            if len(frame) < frames:
                frame.append(page)

            else:
                future = []

                for f in frame:
                    if f in pages[i+1:]:
                        future.append(pages[i+1:].index(f))
                    else:
                        future.append(float('inf'))

                index = future.index(max(future))
                frame[index] = page

        frame_state = frame + ["-"] * (frames - len(frame))

        print(page, "\t", frame_state[0], "\t", frame_state[1], "\t", frame_state[2], "\t", status)

    print("\nTotal Page Faults =", faults)


# reference string
pages = [7,0,1,2,0,3,0,4,2,3,0,3,2,1,2,0,1,7,0,1]
frames = 3

optimal_page_replacement(pages, frames)
```

```python

# LRU Page Replacement

def lru_page_replacement(pages, frames):

    frame = []
    recent = []
    fault = 0

    print("Ref\tFrame1\tFrame2\tFrame3\tStatus")

    for page in pages:

        if page in frame:
            status = "Hit"
            recent.remove(page)
            recent.append(page)

        else:
            fault += 1
            status = "Fault"

            if len(frame) < frames:
                frame.append(page)
                recent.append(page)

            else:
                lru = recent.pop(0)
                index = frame.index(lru)
                frame[index] = page
                recent.append(page)

        frame_state = frame + ["-"] * (frames - len(frame))

        print(page, "\t", frame_state[0], "\t", frame_state[1], "\t", frame_state[2], "\t", status)

    print("\nTotal Page Fault =", fault)


# reference string
pages = [7,0,1,2,0,3,0,4,2,3,0,3,2,1,2,0,1,7,0,1]
frames = 3

lru_page_replacement(pages, frames)
```

```python

# FCFS Disk Scheduling

# input
queue = list(map(int, input("Enter queue (space separated): ").split()))
head = int(input("Enter initial head position: "))

total_distance = 0
current = head

path = [head]

for request in queue:
    distance = abs(request - current)
    total_distance += distance
    current = request
    path.append(current)

# Output Path
print("\nPath:", end=" ")
for p in path:
    print(p, end=" ")

# Total Distance
print("\nTotal Head Movement =", total_distance)
```

```python
#SSTF
def sstf(head, queue):

    requests = queue.copy()
    path = [head]
    total_distance = 0
    current = head

    while requests:
        min_distance = abs(current - requests[0])
        closest = requests[0]

        for r in requests:
            distance = abs(current - r)

            if distance < min_distance:
                min_distance = distance
                closest = r

        total_distance += min_distance

        # head move
        current = closest
        path.append(current)

        # request remove
        requests.remove(closest)

    return path, total_distance

queue = list(map(int, input("Enter Queue: ").split()))
head = int(input("Enter Head Start: "))

path, distance = sstf(head, queue)

print("Path:", *path)
print("Total Distance:", distance)
```

```python
#SCAN
def scan(head, queue):

    queue.sort()

    left = []
    right = []

    for r in queue:
        if r < head:
            left.append(r)
        else:
            right.append(r)

    left.sort(reverse=True)

    path = [head]
    current = head
    total = 0

    # left side
    for r in left:
        total += abs(current - r)
        current = r
        path.append(r)

    # go to 0
    total += abs(current - 0)
    current = 0
    path.append(0)

    # right side
    for r in right:
        total += abs(current - r)
        current = r
        path.append(r)

    return path, total


# INPUT
queue = list(map(int, input("Enter Queue: ").split()))
head = int(input("Enter Head: "))

# PROCESS
path, distance = scan(head, queue)

# OUTPUT
print("Path:", *path)
print("Total Distance:", distance)
```

```python
#CSCAN
def cscan(head, queue, disk_size=200):

    queue.sort()

    left = []
    right = []

    for r in queue:
        if r < head:
            left.append(r)
        else:
            right.append(r)

    path = [head]
    current = head
    total = 0

    # right side service
    for r in right:
        total += abs(current - r)
        current = r
        path.append(r)

    # go to end (199)
    total += abs(current - (disk_size - 1))
    current = disk_size - 1
    path.append(current)

    # jump to 0
    total += abs(current - 0)
    current = 0
    path.append(0)

    # left side service
    for r in left:
        total += abs(current - r)
        current = r
        path.append(r)

    return path, total


# INPUT
queue = list(map(int, input("Enter Queue: ").split()))
head = int(input("Enter Head Start: "))

# PROCESS
path, distance = cscan(head, queue)

# OUTPUT
print("Path:", *path)
print("Total Distance:", distance)
```

```python
#CLOOK
def clook(head, queue):

    queue.sort()

    left = []
    right = []

    for r in queue:
        if r < head:
            left.append(r)
        else:
            right.append(r)

    path = [head]
    current = head
    total = 0

    # right side service
    for r in right:
        total += abs(current - r)
        current = r
        path.append(r)

    # jump to smallest request
    if left:
        total += abs(current - left[0])
        current = left[0]
        path.append(current)

    # serve remaining
    for r in left[1:]:
        total += abs(current - r)
        current = r
        path.append(r)

    return path, total


# INPUT
queue = list(map(int, input("Enter Queue: ").split()))
head = int(input("Enter Head Start: "))

# PROCESS
path, distance = clook(head, queue)

# OUTPUT
print("Path:", *path)
print("Total Distance:", distance)
```
